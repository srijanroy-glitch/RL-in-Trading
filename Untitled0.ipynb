```python
# notebook: portfolio_crypto_analysis.ipynb
# Requirements: pandas numpy scipy matplotlib seaborn yfinance cvxpy

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm, skew, kurtosis, jarque_bera
import yfinance as yf
import os # Import the os module

# --- Ensure directories exist ---
os.makedirs("data/raw", exist_ok=True)
os.makedirs("results", exist_ok=True)

# --- PARAMETERS ---
tickers = {
    'BTC': 'BTC-USD',
    'SPY': 'SPY',
    'AGG': 'AGG'
}
start_date = '2015-01-01'
end_date = '2026-07-27'  # update as needed
trading_days = 252
risk_free_annual = 0.02

# --- DATA LOAD (example using yfinance) ---
def fetch_prices(tickers, start, end):
    data = yf.download(list(tickers.values()), start=start, end=end)['Close']
    data.columns = list(tickers.keys())
    return data

prices = fetch_prices(tickers, start_date, end_date)
prices.to_csv(os.path.join("data/raw", 'prices.csv'))

# --- CLEAN & RETURNS ---
prices = prices.dropna()
logret = np.log(prices / prices.shift(1)).dropna()

# --- ROLLING CORRELATIONS ---
window = 90
rolling_corr_btc_spy = logret['BTC'].rolling(window).corr(logret['SPY'])
rolling_corr_btc_agg = logret['BTC'].rolling(window).corr(logret['AGG'])

# --- PLOT rolling correlations ---
plt.figure(figsize=(10,4))
rolling_corr_btc_spy.plot(label='BTC-SPY (90d)')
rolling_corr_btc_agg.plot(label='BTC-AGG (90d)')
plt.legend(); plt.title('90-day rolling correlations'); plt.show()

# --- PORTFOLIO CONSTRUCTION ---
def portfolio_returns(weights, returns):
    # weights: dict {'BTC':w1, 'SPY':w2, 'AGG':w3}
    w = np.array([weights[t] for t in returns.columns])
    port_ret = returns.dot(w)
    return port_ret

# Example: base 60/40, crypto allocations 0/5/10/20 (reduce SPY proportionally)
def make_weights(crypto_pct):
    # crypto_pct is fraction of total portfolio
    # reduce equities proportionally from 60% equity, 40% bond
    base_equity = 0.6
    base_bond = 0.4
    equity_share = base_equity * (1 - crypto_pct)
    bond_share = base_bond * (1 - crypto_pct)
    return {'BTC': crypto_pct, 'SPY': equity_share, 'AGG': bond_share}

# --- METRICS ---
def annualized_stats(series):
    mean = series.mean() * trading_days
    vol = series.std() * np.sqrt(trading_days)
    sharpe = (mean - risk_free_annual) / vol
    mdd = ( (series + 1).cumprod().cummax() - (series + 1).cumprod() ).max()
    return {'ann_return': mean, 'ann_vol': vol, 'sharpe': sharpe, 'max_drawdown': mdd}

# --- VaR functions ---
def historical_var(returns, alpha=0.95):
    # returns: series of daily returns (decimal)
    q = np.quantile(returns, 1-alpha)
    return -q  # positive number representing loss

def var_cov_var(returns, alpha=0.95):
    mu = returns.mean()
    sigma = returns.std()
    z = norm.ppf(1-alpha)
    var = -(mu + z * sigma)
    return var

def cvar_historical(returns, alpha=0.95):
    q = np.quantile(returns, 1-alpha)
    tail = returns[returns <= q]
    return -tail.mean()

# --- RUN BACKTESTS for periods ---
periods = {
    'full': ('2015-01-01','2026-07-27'),
    'covid_crash': ('2020-02-01','2020-04-30'),
    'bear_2021_22': ('2021-11-01','2022-12-31'),
    'calm_2016_17': ('2016-01-01','2017-12-31')
}

results = {}
for pname, (s,e) in periods.items():
    ret_slice = logret.loc[s:e]
    for cpct in [0.0, 0.05, 0.10, 0.20]:
        w = make_weights(cpct)
        port_series = portfolio_returns(w, ret_slice)
        stats = annualized_stats(port_series)
        # VaR on portfolio and on BTC alone
        hist_var_port = historical_var(port_series, alpha=0.95)
        vc_var_port = var_cov_var(port_series, alpha=0.95)
        hist_var_btc = historical_var(ret_slice['BTC'], alpha=0.95)
        vc_var_btc = var_cov_var(ret_slice['BTC'], alpha=0.95)
        key = (pname, cpct)
        results[key] = {**stats,
                        'hist_var_port_95': hist_var_port,
                        'vc_var_port_95': vc_var_port,
                        'hist_var_btc_95': hist_var_btc,
                        'vc_var_btc_95': vc_var_btc}

# Convert results to DataFrame for reporting
res_df = pd.DataFrame.from_dict(results, orient='index')
res_df.to_csv(os.path.join("results", 'portfolio_metrics.csv'))
```
